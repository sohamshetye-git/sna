```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# Load Shakespeare dataset
url = "https://storage.googleapis.com/download.tensorflow.org/data/shakespeare.txt"
path = tf.keras.utils.get_file("shakespeare.txt", url)

text = open(path, encoding="utf-8").read().lower()[:10000]

# Convert characters into numbers
chars = sorted(set(text))
c2i = {c:i for i,c in enumerate(chars)}
i2c = {i:c for i,c in enumerate(chars)}

# Prepare data
n = 40
X, y = [], []

for i in range(len(text)-n):
    X.append([c2i[c] for c in text[i:i+n]])
    y.append(c2i[text[i+n]])

X = np.array(X)
X = tf.one_hot(X, len(chars))
y = np.array(y)

# Create RNN model
model = Sequential([
    SimpleRNN(32, input_shape=(n, len(chars))),
    Dense(len(chars), activation="softmax")
])

model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy")

# Train
model.fit(X, y, epochs=5, batch_size=64)

# Generate text
generated = text[:n]

for i in range(100):
    x = [[c2i[c] for c in generated[-n:]]]
    x = tf.one_hot(x, len(chars))
    pred = model.predict(x, verbose=0)
    generated += i2c[np.argmax(pred[0])]

print("Generated Shakespeare Text:")
print(generated)
```

```python
import numpy as np
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.utils import pad_sequences

# Load IMDB dataset
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=5000)

# Pad sequences
X_train = pad_sequences(X_train, maxlen=100)
X_test = pad_sequences(X_test, maxlen=100)

# Create RNN model
model = Sequential([
    Embedding(5000, 32),
    SimpleRNN(32),
    Dense(1, activation='sigmoid')
])

# Compile model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Train model
model.fit(X_train, y_train, epochs=3, batch_size=64)

# Evaluate model
loss, accuracy = model.evaluate(X_test, y_test)
print("Test Accuracy:", accuracy)

# Predict sentiment
review = "This movie was excellent and amazing"
word_index = imdb.get_word_index()

words = review.lower().split()
sequence = [word_index.get(w, 2) + 3 for w in words]
sequence = pad_sequences([sequence], maxlen=100)

prediction = model.predict(sequence)

print("Positive" if prediction[0][0] > 0.5 else "Negative")

```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 1. Load IMDB dataset
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=10000)

# 2. Pad sequences to a fixed length
X_train = pad_sequences(X_train, maxlen=200)
X_test = pad_sequences(X_test, maxlen=200)

# 3. Build the LSTM model
model = keras.Sequential([
    keras.layers.Embedding(input_dim=10000, output_dim=64),
    keras.layers.LSTM(64),
    keras.layers.Dense(1, activation="sigmoid")
])

# 4. Compile the model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# 5. Train the model
model.fit(X_train, y_train, epochs=5, batch_size=64)

# 6. Evaluate the model
loss, accuracy = model.evaluate(X_test, y_test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```
