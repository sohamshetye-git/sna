```python
graph = {
    'A':['B','C'],
    'B':['A','D'],
    'C':['A','D'],
    'D':['B','C','E'],
    'E':['D']
}
graph_1 = {
    'A':['B','C'],
    'B':['A','D','E'],
    'C':['A','F','G'],
    'D':['B'],
    'E':['B'],
    'F':['C'],
    'G':['C']
}
graph_2 = {
    'A':['B','C','H'],
    'B':['A','D','E'],
    'C':['A','F','G'],
    'D':['B'],
    'E':['B'],
    'F':['C'],
    'G':['C'],
    'H':['A','I'],
    'I':['H']
}

def structural_equivalence(graph):
    structural = []
    nodes = list(graph.keys())
    for i in range(len(nodes)):
        for j in range(i + 1, len(nodes)):
            node1 = nodes[i]
            node2 = nodes[j]
            if set(graph[node1]) == set(graph[node2]):
                structural.append([node1, node2])
    print("Structurally Equivalent Nodes:")
    print(structural)

print("Graph 1")
structural_equivalence(graph)
print("Graph 2")
structural_equivalence(graph_1)
print("Graph 3")
structural_equivalence(graph_2)

def automorphic(graph):
    automorphic_nodes = []
    nodes = list(graph.keys())
    for i in range(len(nodes)):
        for j in range(i + 1, len(nodes)):
            node1 = nodes[i]
            node2 = nodes[j]
            degree1 = len(graph[node1])
            degree2 = len(graph[node2])
            if degree1 == degree2:
                automorphic_nodes.append([node1, node2])
    print("Automorphic Nodes:")
    print(automorphic_nodes)

print("Graph 1")
automorphic(graph)
print("Graph 2")
automorphic(graph_1)
print("Graph 3")
automorphic(graph_2)

def regular_equivalence(graph):
  nodes = list(graph.keys())
  degrees = {}
  for node in nodes:
    degrees[node] = len(graph[node])
  print("Degrees:")
  print(degrees)
  for i in range(len(nodes)):
    for j in range(i+1,len(nodes)):
      a=nodes[i]
      b=nodes[j]

      a_neigh_deg=sorted(degrees[x] for x in graph[a])
      b_neigh_deg=sorted(degrees[x] for x in graph[b])

      if a_neigh_deg  == b_neigh_deg:
        print("Regular nodes:",a,b)
print("Graph 1")
regular_equivalence(graph)
print("Graph 2")
regular_equivalence(graph_1)
print("Graph 3")
regular_equivalence(graph_2)
```
