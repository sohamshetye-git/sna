#Compute the following node level measures: (i) Density, (ii) Degree, (iii) Reciprocity, (iv) Transitivity, (v) Centralization, (vi) Clustering

#Undirected Graph:Density, Degree, and Reciprocity

```python


nodes = ['A', 'B', 'C', 'D', 'E']

edges = [
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('B', 'D'),
    ('C', 'E')
]

n = len(nodes)
m = len(edges)


# Density
density = (2 * m) / (n * (n - 1))
print("Density =", density)



# Degree
degree = {}

for node in nodes:
    degree[node] = 0

for u, v in edges:
    degree[u] += 1
    degree[v] += 1

print("Degree:")

for node in nodes:
    print(node, "=", degree[node])

# Reciprocity
print("Reciprocity = Not applicable for undirected graph")
```

Directed Graph:Density, Degree, and Reciprocity

```python
# Directed graph

nodes = ['A', 'B', 'C', 'D']

edges = [
    ('A', 'B'),
    ('B', 'A'),
    ('A', 'C'),
    ('B', 'D'),
    ('C', 'D')
]

n = len(nodes)
m = len(edges)


# Density


density = m / (n * (n - 1))

print("Density =", density)

# In-degree and Out-degree


in_degree = {}
out_degree = {}

for node in nodes:
    in_degree[node] = 0
    out_degree[node] = 0

for u, v in edges:
    out_degree[u] += 1
    in_degree[v] += 1

print("\nDegree:")

for node in nodes:
    print(node,
          "In-degree =", in_degree[node],
          "Out-degree =", out_degree[node])


# Reciprocity

reciprocal = 0

for u, v in edges:
    if (v, u) in edges:
        reciprocal += 1

reciprocity = reciprocal / m

print("\nReciprocity =", reciprocity)
```

```python
nodes = ['A', 'B', 'C', 'D', 'E']
edges = [('A','B'), ('A','C'), ('B','C'),
         ('B','D'), ('C','D'), ]

triangles = 0
triples = 0

for a in nodes:
    d = sum(a in e for e in edges)
    triples += d * (d - 1) // 2

for i in range(len(nodes)):
    for j in range(i + 1, len(nodes)):
        for k in range(j + 1, len(nodes)):
            a = nodes[i]
            b = nodes[j]
            c = nodes[k]

            if all(any(set(e) == set(x) for e in edges)
                   for x in [(a,b), (b,c), (a,c)]):
                triangles += 1

print("Triangles =", triangles)
print("Connected triples =", triples)
print("Transitivity =", round(3 * triangles / triples, 2))
```

```python
import matplotlib.pyplot as plt


nodes = ['A', 'B', 'C', 'D', 'E']


edges = [
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('B', 'D'),
    ('C', 'D'),
    ('D', 'E')
]

# Positions
pos = {
    'A': (0, 2),
    'B': (2, 3),
    'C': (2, 1),
    'D': (4, 2),
    'E': (6, 2)
}



triangles = 0
triples = 0

# Count connected triples
for node in nodes:

    neighbors = []

    for u, v in edges:
        if u == node:
            neighbors.append(v)
        elif v == node:
            neighbors.append(u)

    n = len(neighbors)
    triples += n * (n - 1) / 2


# Count triangles
for i in range(len(nodes)):
    for j in range(i + 1, len(nodes)):
        for k in range(j + 1, len(nodes)):

            a = nodes[i]
            b = nodes[j]
            c = nodes[k]

            if ((a, b) in edges or (b, a) in edges) and \
               ((b, c) in edges or (c, b) in edges) and \
               ((a, c) in edges or (c, a) in edges):

                triangles += 1



transitivity = (3 * triangles) / triples

print("Triangles =", triangles)
print("Connected triples =", triples)
print("Transitivity =", transitivity)

for u, v in edges:

    x1, y1 = pos[u]
    x2, y2 = pos[v]

    plt.plot([x1, x2], [y1, y2], 'k-')



for node in nodes:

    x, y = pos[node]

    plt.scatter(x, y, s=700)

    plt.text(
        x, y, node,
        ha='center',
        va='center',
        fontsize=12
    )

plt.title("Undirected Graph")

plt.axis('off')
plt.show()
```

# TRANSITIVITY

```python
import matplotlib.pyplot as plt


nodes = ['A', 'B', 'C', 'D', 'E']


edges = [
    ('A', 'B'),
    ('B', 'C'),
    ('A', 'C'),

    ('B', 'D'),
    ('D', 'E'),
    ('B', 'E'),   #

    ('A', 'D'),
    ('A', 'E')
]




total = 0
closed = 0

for a, b in edges:
    for x, c in edges:


        if b == x and a != c:

            total += 1


            if (a, c) in edges:
                closed += 1

if total > 0:
    transitivity = closed / total
else:
    transitivity = 0

print("Directed triples =", total)
print("Closed triples =", closed)
print("Transitivity =", transitivity)

pos = {
    'A': (0, 2),
    'B': (2, 3),
    'C': (4, 3),
    'D': (2, 1),
    'E': (4, 1)
}


for u, v in edges:

    x1, y1 = pos[u]
    x2, y2 = pos[v]

    plt.arrow(
        x1, y1,
        x2-x1, y2-y1,
        head_width=0.12,
        length_includes_head=True
    )


for node in nodes:

    x, y = pos[node]

    plt.scatter(x, y, s=700)

    plt.text(
        x, y, node,
        ha='center',
        va='center',
        fontsize=12
    )

plt.title("Directed Graph")

plt.axis('off')
plt.show()
```

```python
nodes = ['A', 'B', 'C', 'D', 'E']
edges = [
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('B', 'E'),
    ('C', 'E'),
    ('C', 'D'),
    ('D', 'E'),
]

for n in nodes:
    dist = {n: 0}
    queue = [n]

    while queue:
        u = queue.pop(0)
        for a, b in edges:
            v = b if a == u else a if b == u else None
            if v is not None and v not in dist:
                dist[v] = dist[u] + 1
                queue.append(v)

    centrality = (len(nodes)-1) / sum(dist.values())
    print(n, "Closeness Centrality =", round(centrality, 2))
```

```python
import matplotlib.pyplot as plt


nodes = ['A', 'B', 'C', 'D', 'E', 'F']

edges = [
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('B', 'D'),
    ('C', 'E'),
    ('D', 'E'),
    ('D', 'F'),
    ('E', 'F')
]



graph = {node: [] for node in nodes}

for u, v in edges:
    graph[u].append(v)
    graph[v].append(u)


degree_centrality = {}

for node in nodes:
    degree = len(graph[node])
    degree_centrality[node] = degree / (len(nodes) - 1)
```

```python
def shortest_paths(start):

    distance = {node: -1 for node in nodes}
    distance[start] = 0

    queue = [start]

    while queue:

        current = queue.pop(0)

        for neighbor in graph[current]:

            if distance[neighbor] == -1:
                distance[neighbor] = distance[current] + 1
                queue.append(neighbor)

    return distance
```

CLOSENESS CENTRALITY
#Formula = (n-1) / sum(distances)

```python



closeness = {}

for node in nodes:

    distances = shortest_paths(node)

    total_distance = sum(distances.values())

    closeness[node] = (len(nodes) - 1) / total_distance
```

```python
nodes = ['A', 'B', 'C', 'D']
edges = [('A','B'), ('B','C'), ('C','D')]

for v in nodes:
    count = 0

    for a in nodes:
        for b in nodes:
            if a < b and a != v and b != v:
                # Check if v lies between a and b
                dist = {a: 0}
                queue = [a]

                while queue:
                    u = queue.pop(0)
                    for x, y in edges:
                        w = y if x == u else x if y == u else None
                        if w is not None and w not in dist:
                            dist[w] = dist[u] + 1
                            queue.append(w)

                d1 = dist.get(b, 999)
                d2 = dist.get(v, 999)

                if d1 == d2 + 1:
                    count += 1

    print(v, "Betweenness Centrality =", count)
```

```python
nodes = ['A', 'B', 'C', 'D','E']
edges = [('A','B'), ('A','C'), ('A','D'),('A','E'),
         ('D','E')]

for n in nodes:
    nei = [v if u == n else u for u,v in edges if u == n or v == n]
    k = len(nei)
    e = sum(1 for i in range(k) for j in range(i+1,k)
            if (nei[i],nei[j]) in edges or (nei[j],nei[i]) in edges)
    c = 2*e/(k*(k-1)) if k > 1 else 0
    print(n, "=", round(c, 2))
```

```python
import matplotlib.pyplot as plt


nodes = ['A', 'B', 'C', 'D', 'E']


edges = [
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('B', 'D'),
    ('C', 'D'),
    ('C', 'E')
]





for node in nodes:

    # neighbors
    neighbors = []

    for u, v in edges:
        if u == node:
            neighbors.append(v)
        elif v == node:
            neighbors.append(u)

    k = len(neighbors)

    # edges between neighbors
    e = 0

    for i in range(k):
        for j in range(i + 1, k):

            if (neighbors[i], neighbors[j]) in edges or \
               (neighbors[j], neighbors[i]) in edges:

                e += 1

    # clustering
    if k < 2:
        clustering = 0
    else:
        clustering = (2 * e) / (k * (k - 1))

    print(node, "=", round(clustering, 2))





```
