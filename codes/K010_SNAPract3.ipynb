AIM: Represent a network using single mode and double mode data. Compute the adjacency matrix, degree diagonal matrix, Laplacian matrix, transition probability matrix, and incidence matrix.

```python
import numpy as np
import pandas as pd


df = pd.read_csv('/content/Friendship.csv', index_col=0)

print("Adjacency Matrix:")
print(df)

# Convert to numpy array
A = df.values

# Node names
labels = df.index.tolist()

print("\nArray:")
print(A)
```

# 1. Degree Matrix

```python

degree_values = np.sum(A, axis=1)

D = np.diag(degree_values)

print("\nDegree Matrix:")
print(D)
```

# 2. Laplacian Matrix

```python

L = D - A

print("\nLaplacian Matrix:")
print(L)
```

# 3. Transition Probability Matrix



```python
P = np.divide(
    A,
    degree_values[:, np.newaxis],
    out=np.zeros_like(A, dtype=float),
    where=degree_values[:, np.newaxis] != 0
)

print("\nTransition Probability Matrix:")
print(P)
```

# 4. Edge List

```python
edges = []

for i in range(len(labels)):

    for j in range(i + 1, len(labels)):

        if A[i, j] == 1:

            edges.append((i, j))


print("\nEdges:")

for u, v in edges:

    print(labels[u], "-", labels[v])

```

# 5. Incidence Matrix

```python
I = np.zeros(
    (len(labels), len(edges)),
    dtype=int
)

for k, (u, v) in enumerate(edges):

    I[u, k] = 1
    I[v, k] = 1


print("\nIncidence Matrix:")
print(I)
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv('/content/Two_Mode_Student_Course.csv', index_col=0)

B = df.values
students = df.index.tolist()
courses = df.columns.tolist()

print("Two-Mode Matrix:")
print(df)

# Adjacency Matrix
A = np.block([
    [np.zeros((len(students), len(students)), dtype=int), B],
    [B.T, np.zeros((len(courses), len(courses)), dtype=int)]
])

nodes = students + courses

print("\nAdjacency Matrix:")
print(pd.DataFrame(A, index=nodes, columns=nodes))


```

```python
# Degree Diagonal Matrix
degree = np.sum(A, axis=1)
D = np.diag(degree)

print("\nDegree Diagonal Matrix:")
print(pd.DataFrame(D, index=nodes, columns=nodes))
```

```python

# Laplacian Matrix
L = D - A

print("\nLaplacian Matrix:")
print(pd.DataFrame(L, index=nodes, columns=nodes))
```

```python
# Transition Probability Matrix
P = np.divide(
    A,
    degree[:, np.newaxis],
    out=np.zeros_like(A, dtype=float),
    where=degree[:, np.newaxis] != 0
)

print("\nTransition Probability Matrix:")
print(pd.DataFrame(P, index=nodes, columns=nodes))

```

```python
# Edge List
edges = []

for i in range(len(students)):
    for j in range(len(courses)):
        if B[i, j] == 1:
            edges.append((students[i], courses[j]))

print("\nEdge List:")
for edge in edges:
    print(edge[0], "-", edge[1])
```

```python
# Incidence Matrix
I = np.zeros((len(nodes), len(edges)), dtype=int)

for k, (student, course) in enumerate(edges):
    I[students.index(student), k] = 1
    I[len(students) + courses.index(course), k] = 1

print("\nIncidence Matrix:")
print(pd.DataFrame(
    I,
    index=nodes,
    columns=[f"E{i+1}" for i in range(len(edges))]
))
```
