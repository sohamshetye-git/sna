```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()
G.add_edges_from([('A','C'), ('A','D'), ('B','C'), ('B','D')])

# Detect bipartite groups
color = nx.bipartite.color(G)

# Check if bipartite
if nx.is_bipartite(G):
    group1 = [n for n in G if color[n] == 0]
    group2 = [n for n in G if color[n] == 1]

    pos = nx.bipartite_layout(G, group1)
    node_colors = ['red' if color[n] == 0 else 'skyblue' for n in G]

    nx.draw(G, pos, with_labels=True,
            node_color=node_colors, node_size=1500)
    plt.show()
else:
    print("Graph is not bipartite")
```

```python
nodes = ['A', 'B', 'C', 'D']
edges = [('A', 'C'), ('A', 'D'), ('B', 'C'), ('B', 'D')]

color = {}
bipartite = True

for start in nodes:
    if start not in color:
        color[start] = 0
        queue = [start]

        while queue:
            u = queue.pop(0)

            for a, b in edges:
                if a == u:
                    v = b
                elif b == u:
                    v = a
                else:
                    continue

                if v not in color:
                    color[v] = 1 - color[u]
                    queue.append(v)
                elif color[v] == color[u]:
                    bipartite = False

print("Bipartite:", bipartite)
```

```python
import matplotlib.pyplot as plt
from collections import deque


def check_bipartite(graph):
    color = {}
    conflict = None

    for start in graph:

        if start not in color:
            color[start] = 0
            queue = deque([start])

            while queue:
                node = queue.popleft()

                for neighbor in graph[node]:

                    if neighbor not in color:
                        color[neighbor] = 1 - color[node]
                        queue.append(neighbor)

                    elif color[neighbor] == color[node]:
                        if conflict is None:
                            conflict = (node, neighbor)

    if conflict is None:
        return True, color, None

    return False, color, conflict


def draw_graph(graph, position, title):

    result, color, conflict = check_bipartite(graph)

    plt.figure(figsize=(6, 5))

    for node in graph:
        for neighbor in graph[node]:

            if node < neighbor:

                x1, y1 = position[node]
                x2, y2 = position[neighbor]

                plt.plot(
                    [x1, x2],
                    [y1, y2],
                    color="black",
                    linewidth=1.5
                )

    for node in graph:

        x, y = position[node]

        if color[node] == 0:
            node_color = "skyblue"
        else:
            node_color = "lightcoral"

        plt.scatter(
            x,
            y,
            s=800,
            color=node_color,
            edgecolors="black",
            zorder=2
        )

        plt.text(
            x,
            y,
            str(node),
            ha="center",
            va="center",
            fontsize=10
        )

    if result:

        plt.title(
            title + " - BIPARTITE",
            color="green"
        )

    else:

        plt.title(
            title + " - NOT BIPARTITE",
            color="red"
        )

    plt.axis("off")
    plt.tight_layout()
    plt.show()

    if result:

        group1 = []
        group2 = []

        for node in color:

            if color[node] == 0:
                group1.append(node)
            else:
                group2.append(node)

        print(title, "-> BIPARTITE")
        print("Group 1:", group1)
        print("Group 2:", group2)

    else:

        print(title, "-> NOT BIPARTITE")
        print(
            "Conflict nodes:",
            conflict[0],
            "and",
            conflict[1]
        )

    print()


G1 = {
    1: [4, 2, 8],
    2: [3, 1, 7],
    3: [4, 2, 6],
    4: [3, 1, 5],
    5: [4, 6, 8],
    6: [3, 5, 7],
    7: [2, 6, 8],
    8: [1, 5, 7]
}

pos1 = {
    4: (-1.5, 1),
    3: (1.5, 1),
    2: (1.5, -1),
    1: (-1.5, -1),
    5: (-0.6, 0.5),
    6: (0.6, 0.5),
    7: (0.6, -0.5),
    8: (-0.6, -0.5)
}


G2 = {
    1: [2, 3, 4],
    2: [1, 5, 6],
    3: [1, 6],
    4: [1, 7],
    5: [2],
    6: [2, 3],
    7: [4]
}

pos2 = {
    1: (0, 1.5),
    2: (-1, 0),
    3: (0.2, 0),
    4: (1.2, 0),
    5: (-1.5, -1.3),
    6: (0, -1.3),
    7: (1.2, -1.3)
}


G3 = {
    2: [3, 1],
    3: [2, 6],
    1: [2, 5],
    6: [3, 5],
    5: [1, 6],
    7: [3]
}

pos3 = {
    2: (0, 1.5),
    3: (-1, 0.5),
    1: (1, 0.5),
    7: (-1.5, -0.8),
    6: (0.5, -0.8),
    5: (1.5, -1.5)
}


G4 = {
    2: [1, 3, 4],
    1: [2, 3],
    3: [2, 1, 4, 5, 6],
    4: [2, 3, 5],
    5: [3, 4],
    6: [3]
}

pos4 = {
    2: (0, 1.3),
    1: (-1.4, 0.2),
    3: (1, 0.2),
    4: (0, -1.2),
    5: (-1.2, -1.5),
    6: (1.7, -0.6)
}


G5 = {
    5: [1, 4],
    6: [4],
    2: [1, 4],
    3: [1, 4],
    4: [2, 3, 5, 6],
    1: [2, 3, 5]
}

pos5 = {
    5: (-0.8, 1.3),
    6: (0.8, 1.3),
    2: (-1.5, 0),
    4: (0, 0),
    3: (1.5, 0),
    1: (0, -1.5)
}


draw_graph(G1, pos1, "Graph 1")
draw_graph(G2, pos2, "Graph 2")
draw_graph(G3, pos3, "Graph 3")
draw_graph(G4, pos4, "Graph 4")
draw_graph(G5, pos5, "Graph 5")
```
